```python
def fetch_url(url, timeout=10):
    try:
        res = requests.get(url, timeout=timeout)
        res.raise_for_status()
        print(res.status_code)
        return res.text
    except requests.exceptions.RequestException as e:
        print(f"Error fetching {url}: {e}")
        return None

html = fetch_url("https://www.reddit.com")
print(html[0:200])
```

```python
soup = BeautifulSoup(html, "lxml")
print(soup.prettify()[0:500])
```

```python
element = soup.find("div", class_="subgrid-container")
# print(element[:200])

links = soup.find_all("a")
for link in links[0:5]:
    # print(link.text)
    print(link['href'])
```

```python
html = requests.get("https://quotes.toscrape.com").text
soup = BeautifulSoup(html,"lxml")
# print(soup.prettify()[:500])

quotes = soup.find_all("div", class_="quote")
for q in quotes:
    quote_text = q.find("span",class_="text").text
    print("quote: ",quote_text)
    quote_author = q.find("small", class_="author").text
    print("author: ", quote_author)
    tags = q.find_all("a", class_="tag")
    print("tags: ",end=' ')
    for tag in tags:
        print(tag.text, end = " ")
    print("\n\n")

next_page = soup.find("li",class_="next")
link = next_page.find("a")
print(link["href"])
```
