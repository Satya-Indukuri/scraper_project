# Object-Oriented Programming (OOP)

```python
#Basic Class
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def speak(self):
        return f"{self.name} says woof"
#Instance
d = Dog("Simba", 2)
print(d.name,d.age)
print(d.speak())
```

```python
import math

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance_to_origin(self):
        return math.hypot(self.x,self.y)

p = Point(3,4)
assert p.distance_to_origin() == 5, "Incorrect Result"
print(p.distance_to_origin())
```

```python
# Instance vs Class variables
# Instance variables are per-object.
# Class variables are shared across instances.
class Counter:
    total = 0 #class variable
    
    def __init__(self):
        Counter.total += 1
        self.id = Counter.total #instance variable
        
c1 = Counter()
c2 = Counter()
print(c1.id, c2.id, Counter.total)
```

```python
#access specifiers: public(), protected(_), private(__)
class Example:
    def __init__(self, x, y):
        self.public_var = x
        self._protected_var = y
        self.__private_var = x*y

e = Example(2,3)
print(e.public_var, e._protected_var, e._Example__private_var) #syntax to access private attribute - _ClassName__attr
```

```python
#setters and getters
class MyClass:
    def __init__(self, value):
        self._value = value
    @property          #acts like a getter
    def value(self): 
        return self._value

    @value.setter    #acts like a setter
    def value(self, value):
        if(value>0):
            self._value = value
        else:
            print("Value cannot be negative")

obj = MyClass(10)
print(obj.value)
obj.value = 20 
print(obj.value)
obj.value = -5 
print(obj.value)
```

```python
#Inheritance and Polymorphism: subclass extends or customizes parent behavior.
class Animal:
    def speak(self):
        return "sound"

class Dog(Animal):
    def speak(self):
        return "woof"

class Cat(Animal):
    def speak(self):
        return "meow"

for cls in (Animal, Cat, Dog):
    print(cls.__name__, "->", cls().speak())
```

```python
#Composition
class Engine:
    def start(self):
        return "vroom"
class Car:
    def __init__(self):
        self.engine = Engine()   #composition
    def drive(self):
        return self.engine.start()

car = Car()
print(car.drive())
```

```python
#Inheritance
class Vehicle:
    def __init__(self,brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    def display_info(self):
        print(f"brand: {self.brand}, model:{self.model}, year:{self.year}")

v1 = Vehicle("Toyota","Camry",2025)
v1.display_info()

class Car(Vehicle):
    def __init__(self, brand, model, year, num_doors):
        super().__init__(brand, model, year)
        self.num_doors = num_doors

    def display_info(self):
        print(f"Car: {self.brand} {self.model} ({self.year}) - Doors: {self.num_doors}")

c1 = Car("Honda", "Civic", 2022, 4)
c1.display_info()

class Motorcycle(Vehicle):
    def __init__(self, brand, model, year, has_sidecar):
        super().__init__(brand, model, year)
        self.has_sidecar = has_sidecar

    def display_info(self):
        sidecar_status = "with sidecar" if self.has_sidecar else "no sidecar"
        print(f"Motorcycle: {self.brand} {self.model} ({self.year}) - {sidecar_status}")

m1 = Motorcycle("RoyalEnfield","GT650", 2024, True)
m1.display_info()

for vehicle in (v1,c1,m1):
    vehicle.display_info()
```

```python
#Encapsulation
class BankAccount():
    def __init__(self, account_number, balance=0):
        self.__account_number = account_number
        self.__balance = balance

    def deposit(self, amount):
        if amount<0:
            print("Deposit amount must be positive.")
        else:
            self.__balance += amount
            print(f"Deposited {amount}. New balance: {self.__balance}")

    def withdraw(self, amount):
        if amount>self.__balance:
            print("Insufficient funds")
        elif amount<=0:
            print("Withdraw amount must be positive.")
        else:
            self.__balance -= amount
            print(f"Withdrew {amount}. New balance: {self.__balance}")

    def get_balance(self):
        return self.__balance
        # print(f"Balance: {self.__balance}")

# Create accounts
acc1 = BankAccount("ACC123", 1000)
acc2 = BankAccount("ACC456", 500)

print(acc1._BankAccount__account_number) #to access the private attributes outside of class, syntax - _className__attribteName
print(acc2. _BankAccount__balance)

# Deposit
acc1.deposit(200)
acc2.deposit(-50)

# Withdraw
acc1.withdraw(300)
acc2.withdraw(600)

# Access balance safely
print(acc1.get_balance())
```

```python
#Polymorphism
class Shape:
    def __init__(self):
        pass
    def area(self):
        return 0
    def display(self):
        print("This is a shape")

class Triangle(Shape):
    def __init__(self, height, base):
        super().__init__()
        self.height = height
        self.base = base
    def area(self):
        return 0.5*self.height*self.base
    def display(self):
        print("This is a triangle")

class Rectangle(Shape):
    def __init__(self, breadth, length):
        super().__init__()
        self.l = length
        self.b = breadth
    def area(self):
        return self.l*self.b
    def display(self):
        print("This is a Rectangle")

def print_area(shape_obj):
    shape_obj.display()
    print("Area:", shape_obj.area())

shapes = [
    Triangle(5, 6),
    Rectangle(6, 4)
]

for s in shapes:
    print_area(s)
```

```python
#Abstarction
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def make_payment(self, amount):
        pass
    def payment_summary(self):
        print("payment processed succesfully")

class CreditCardPayment(Payment):
    def make_payment(self, amount):
        print(f"Paid ${amount} using Credit Card.")

class PayPalPayment(Payment):
    def make_payment(self, amount):
        print(f"Paid ${amount} using PayPal.")

def process_payment(payment_obj, amount):
    payment_obj.make_payment(amount)
    payment_obj.payment_summary()

# Creating objects of subclasses
credit = CreditCardPayment()
paypal = PayPalPayment()

process_payment(credit, 100)
process_payment(paypal, 200)
```

```python
class Book:
    def __init__(self, title, author, available = True):
        self.title = title
        self.author = author
        self.available = available
        
    def borrow(self):
        if not available:
            print("book not available")
            return
        else:
            self.available = False
            
    def return_book(self):
        self.available = True
        
    def info(self):
        print(f"title: {self.title} \nauthor:{self.author} \navailable:{self.available} ")

class Member:
    def __init__(self, name):
        self.name = name
        self.borrowed_books = []

    def borrow_book(self, book):
        if book.available == False:
            print("Book already borrowed.")
        else:
            book.available = False
            self.borrowed_books.append(book)

    def return_book(self, book):
        book.available = True
        self.borrowed_books.remove(book)

    def list_books(self):
        for book in self.borrowed_books:
            print(f"{book.title} by {book.author}")

class Library:
    def __init__(self):
        self.books = []
        self.members = []

    def add_book(self, book):
        self.books.append(book)

    def add_member(self, member):
        self.members.append(member)

    def show_all_books(self):
        for book in self.books:
            availability = "yes" if book.available==True else "no"
            print(f"{book.title} by {book.author}  availability: {availability} ")

    def lend_book(self, book_title, member_name):
        book = None
        member = None
        for b in self.books:
            if b.title == book_title:
                book = b
                break
        if book == None:
            print("Book not found")
            return
        for m in self.members:
            if m.name == member_name:
                member = m
                break
        if member == None:
            print("member not found")
            return

        if book.available == False:
            print("book not available")
        else:
            member.borrow_book(book)
            print(f"{member_name} borrowed {book_title}")

    def return_book(self, book_title, member_name):
        book = None
        member = None
        for b in self.books:
            if b.title == book_title:
                book = b
                break
        if book == None:
            print("Book not found")
            return
        for m in self.members:
            if m.name == member_name:
                member = m
                break
        if member == None:
            print("member not found")
            return
        member.return_book(book)
        print(f"{member_name} returned {book_title}")
```

```python
lib = Library()
b1 = Book("1984", "George Orwell")
b2 = Book("To Kill a Mockingbird", "Harper Lee")
m1 = Member("Satya")

lib.add_book(b1)
lib.add_book(b2)
lib.add_member(m1)

lib.show_all_books()
lib.lend_book("1984", "Satya")
m1.list_books()
lib.show_all_books()
lib.return_book("1984", "Satya")
lib.show_all_books()
```
